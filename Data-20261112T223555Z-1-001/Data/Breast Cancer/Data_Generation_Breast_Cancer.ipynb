```python
# DATA GENERATION FOR BREAST CANCER
from mostlyai.sdk import MostlyAI
import pandas as pd
from google.colab import files


# Step 1 — Upload and load extended dataset
df_full = pd.read_csv("cancer-risk-factors-extended-1.csv")

# Step 2 — Extract breast rows (keep ALL features)
df_breast = df_full[df_full["Cancer_Type"] == "Breast"].copy()
df_breast.drop(columns=["Patient_ID"], inplace=True, errors='ignore')

print(f"breast rows: {df_breast.shape}")
print(df_breast["Risk_Level"].value_counts())

# Step 3 — Train MOSTLY AI on breast rows only
mostly = MostlyAI()
g = mostly.train(
    name='breast_cancer',
    data=df_breast,
    start=True,
    wait=True
)

# Step 4 — Generate
n = 3000
sd = mostly.generate(g, size=n)
df_synthetic = sd.data()

print(f"\nSynthetic: {df_synthetic.shape}")
print(df_synthetic["Risk_Level"].value_counts())

# Step 5 — Combine
df_breast_extended = pd.concat(
    [df_breast, df_synthetic],
    ignore_index=True
)

print(f"\nFinal: {df_breast_extended.shape}")
print(df_breast_extended["Risk_Level"].value_counts())
print(df_breast_extended["Risk_Level"].value_counts(normalize=True).round(3))

# Step 6 — Download
df_breast_extended.to_csv("breast_cancer_extended.csv", index=False)
files.download("breast_cancer_extended.csv")
print("Done")
```
