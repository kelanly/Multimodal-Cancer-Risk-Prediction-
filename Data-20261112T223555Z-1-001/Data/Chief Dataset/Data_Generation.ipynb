```python
from google.colab import files
uploaded = files.upload()  # select cancer-risk-factors.csv from your computer

import pandas as pd
df = pd.read_csv("cancer-risk-factors-shaved-60-20-20.csv")
print(f"Loaded: {df.shape}")
print(df["Risk_Level"].value_counts())
```

```python
from mostlyai.sdk import MostlyAI

mostly = MostlyAI()
g = mostly.train(
    name='cancer-risk-factors-shaved-60-20-20-extended',
    data=df,
    start=True,
    wait=True,
)

# Step 4 — Generate
sd = mostly.generate(g, size=3000)
df_synthetic = sd.data()
print(f"Synthetic: {df_synthetic.shape}")

# Step 5 — Combine
df_extended = pd.concat([df, df_synthetic], ignore_index=True)
print(f"Extended: {df_extended.shape}")
print(df_extended["Risk_Level"].value_counts())

# Step 6 — Save and download
df_extended.to_csv("cancer-risk-factors-shaved-60-20-20-extended.csv", index=False)
files.download("cancer-risk-factors-shaved-60-20-20-extended.csv")
print("Done")
```

```python
# Sanity check on generated data
print("=== Shape ===")
print(df_extended.shape)

print("\n=== Risk Level Distribution ===")
print(df_extended["Risk_Level"].value_counts())

print("\n=== Basic Stats ===")
print(df_extended.describe())

print("\n=== Sample Rows ===")
print(df_extended.tail(5))  # last 5 rows should be synthetic
```

```python
# compare with previous dataset
# Compare original vs extended distributions
import pandas as pd

print("=== Original ===")
print(df["Cancer_Type"].value_counts())
print(df["Risk_Level"].value_counts())

print("\n=== Extended ===")
print(df_extended["Cancer_Type"].value_counts())
print(df_extended["Risk_Level"].value_counts())
```
