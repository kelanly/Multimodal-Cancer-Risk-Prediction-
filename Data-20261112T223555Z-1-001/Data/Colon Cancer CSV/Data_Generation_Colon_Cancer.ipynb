```python
# DATA GENERATION FOR COLON CANCER
from mostlyai.sdk import MostlyAI
import pandas as pd
from google.colab import files


# Step 1 — Upload and load extended dataset
df_full = pd.read_csv("cancer-risk-factors-extended-1.csv")

# Step 2 — Extract Colon rows (keep ALL features)
df_colon = df_full[df_full["Cancer_Type"] == "Colon"].copy()
df_colon.drop(columns=["Patient_ID"], inplace=True, errors='ignore')

print(f"Colon rows: {df_colon.shape}")
print(df_colon["Risk_Level"].value_counts())

# Step 3 — Train MOSTLY AI on colon rows only
mostly = MostlyAI()
g = mostly.train(
    name='colon_cancer',
    data=df_colon,
    start=True,
    wait=True
)

# Step 4 — Generate
n = 3000
sd = mostly.generate(g, size=n)
df_synthetic = sd.data()

print(f"\nSynthetic: {df_synthetic.shape}")
print(df_synthetic["Risk_Level"].value_counts())

# Step 5 — Combine
df_colon_extended = pd.concat(
    [df_colon, df_synthetic],
    ignore_index=True
)

print(f"\nFinal: {df_colon_extended.shape}")
print(df_colon_extended["Risk_Level"].value_counts())
print(df_colon_extended["Risk_Level"].value_counts(normalize=True).round(3))

# Step 6 — Download
df_colon_extended.to_csv("colon_cancer_extended.csv", index=False)
files.download("colon_cancer_extended.csv")
print("Done")
```

```python
# Sanity check on generated data
print("=== Shape ===")
print(df_colon_extended.shape)

print("\n=== Risk Level Distribution ===")
print(df_colon_extended["Risk_Level"].value_counts())

print("\n=== Basic Stats ===")
print(df_colon_extended.describe())

print("\n=== Sample Rows ===")
print(df_colon_extended.tail(5))  # last 5 rows should be synthetic
```
