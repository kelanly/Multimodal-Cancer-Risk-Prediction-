```python
# DATA GENERATION FOR SKIN CANCER
from mostlyai.sdk import MostlyAI
import pandas as pd
from google.colab import files


# Step 1 — Upload and load extended dataset
df_full = pd.read_csv("cancer-risk-factors-extended-1.csv")

# Step 2 — Extract Skin rows (keep ALL features)
df_skin = df_full[df_full["Cancer_Type"] == "Skin"].copy()
df_skin.drop(columns=["Patient_ID"], inplace=True, errors='ignore')

print(f"Skin rows: {df_skin.shape}")
print(df_skin["Risk_Level"].value_counts())

# Step 3 — Train MOSTLY AI on skin rows only
mostly = MostlyAI()
g = mostly.train(
    name='skin_cancer',
    data=df_skin,
    start=True,
    wait=True
)

# Step 4 — Generate
n = 3000
sd = mostly.generate(g, size=n)
df_synthetic = sd.data()

print(f"\nSynthetic: {df_synthetic.shape}")
print(df_synthetic["Risk_Level"].value_counts())

# Step 5 — Combine
df_skin_extended = pd.concat(
    [df_skin, df_synthetic],
    ignore_index=True
)

print(f"\nFinal: {df_skin_extended.shape}")
print(df_skin_extended["Risk_Level"].value_counts())
print(df_skin_extended["Risk_Level"].value_counts(normalize=True).round(3))

# Step 6 — Download
df_skin_extended.to_csv("skin_cancer_extended.csv", index=False)
files.download("skin_cancer_extended.csv")
print("Done")
```
