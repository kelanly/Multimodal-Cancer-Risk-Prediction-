## Skew
**Skew** is represented by '$a$'.

```
If a is negative => left skew
Else if a is positive => right skew
Else: No skew
```
Formula to calculate skew:
$$
a = \frac{3(Mean-Median)}{Standard Deviation}
$$

## Scale

Approximates to Standard Deviation. This is one of the input scipy's skewnorm() demands.
